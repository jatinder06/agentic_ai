```python
os.environ['HF_TOKEN']=os.getenv("HF_TOKEN")
os.environ['GEMINI_API_KEY']=os.getenv("GEMINI_API_KEY")
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
embeddings=HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
```

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
model=GoogleGenerativeAIEmbeddings(model='models/embedding-001')
```

```python
embeddings=model.embed_query("What is the capital of France?")
print(embeddings)
```

```python
text = "This is a test text for embedding."
# Get the embedding for the text
embedding = embeddings.embed_query(text)
print(embedding)
print(len(embedding))  # Should print the dimension of the embedding vector, typically 384 for this model
```
